```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv( 'Sample - Superstore.csv' , encoding='Windows-1252' )


df.head()
```

```python
print("--- Data Info ---")
df.info()

print("\n--- Data Shape (Rows, Columns) ---")
print(df.shape)
```

```python
region_analysis = df.groupby('Region') [['Sales', 'Profit']].sum()
print("--- Region-wise Performance ---")
print(region_analysis)
```

```python
region_analysis['Profit'].plot(kind='bar', color='teal', figsize=(8,4))
import matplotlib.pyplot as plt
plt.title('Profit by Region')
plt.ylabel('Profit ($)')
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df,x='Sales', y='Profit', hue='Discount', palette='viridis',alpha=0.8)

plt.title('Sales vs Profit Analysis(Colored by Discount)', fontsize=14)
plt.xlabel('Sales ($)', fontsize=12)
plt.ylabel('Profit ($)', fontsize=12)
plt.axhline(0, color='red', linestyle='--', linewidth=1)

plt.tight_layout()
plt.show()
```
